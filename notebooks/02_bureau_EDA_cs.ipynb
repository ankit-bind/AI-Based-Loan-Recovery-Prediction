# Exploratory Data Analysis (EDA) — bureau.csv

# 1 ) **Introduction**

## 1.1 Bureau Dataset Overview

The bureau dataset contains historical credit records of loan applicants collected from external financial institutions. Each record represents a previous or ongoing credit account associated with a borrower, providing additional information about borrower financial activity and repayment behavior.

## 1.2 Dataset Information

Dataset Name: Home Credit Default Risk — bureau.csv

The dataset includes important credit-related information such as:
- Credit status
- Credit type
- Outstanding debt
- Overdue history
- Credit prolongation
- Credit timelines
- External borrowing behavior

Multiple bureau records may exist for a single applicant, making aggregation-based feature engineering important.

## 1.3 Objective of EDA

The objective of this analysis is to understand borrower credit behavior, analyze debt exposure, study overdue patterns, identify repayment-related risks, and generate useful bureau-based insights for loan recovery prediction modeling.


## 1.4 Dataset Feature Description

The bureau dataset contains historical credit information
of borrowers obtained from external financial institutions.

The dataset includes important credit-related features such as:

- Credit account status
- Credit type and currency
- Outstanding debt amount
- Overdue information
- Credit prolongation history
- Credit timelines and update history
- External borrower financial activity

Each borrower may have multiple bureau records,
making the dataset useful for
aggregation-based feature engineering
and repayment risk analysis.

# 3 ) **Missing Value Analysis**

```python
# MISSING VALUE ANALYSIS
# Total missing values
total = bureau.isnull().sum()

# Missing percentage
percent = (bureau.isnull().sum() / len(bureau)) * 100

# Combine results
missing_df = pd.concat([total, percent],axis=1,keys=['Total Missing','Missing Ratio (%)'])

# Sort values
missing_df = missing_df.sort_values(by='Missing Ratio (%)',ascending=False)

# Display top missing columns
missing_df.head(15)
```

# 4 ) **Bureau Credit Analysis**

 * AMT_ANNUITY and AMT_CREDIT_MAX_OVERDUE
   contain very high missing values,
   exceeding 65% of the dataset.

 * DAYS_ENDDATE_FACT and AMT_CREDIT_SUM_LIMIT
  also contain substantial missing information.

 * Most core credit-related features
   have very low or no missing values,
   making them reliable for feature engineering.

 * Missing value handling will be important
   before aggregation and model training.


## 4.1 CREDIT STATUS ANALYSIS

```python

# CREDIT STATUS ANALYSIS
# Credit status distribution
plt.figure(figsize=(10,5))
sns.countplot(y='CREDIT_ACTIVE',data=bureau,order=bureau['CREDIT_ACTIVE'].value_counts().index)
plt.title('Distribution of Credit Status')
plt.xlabel('Count')
plt.ylabel('Credit Status')
plt.show()
```



 * Closed credits dominate the bureau dataset,
   indicating that many borrowers have completed
   previous loans successfully.

 * Active credits also represent a large portion
   of records,
   showing ongoing financial obligations
   for many applicants.

* Sold and Bad Debt records are very small,
   but may indicate higher financial risk borrowers.

 * CREDIT_ACTIVE can be an important feature
   for understanding borrower credit behavior
   and repayment stability.

## 4.2  CREDIT TYPE DISTRIBUTION

```python

# CREDIT TYPE DISTRIBUTION
plt.figure(figsize=(12,6))
sns.countplot(y='CREDIT_TYPE',data=bureau,order=bureau['CREDIT_TYPE'].value_counts().head(10).index)
plt.title('Top Credit Types in Bureau Dataset')
plt.xlabel('Count')
plt.ylabel('Credit Type')
plt.show()
```



 * Consumer credit dominates the bureau dataset,
   indicating that most borrowers rely on personal loans.

 * Credit card usage is also significant,
   showing active revolving credit behavior
   among applicants.

 * Mortgage and Car loan records are relatively low,
   suggesting fewer long-term secured loans.

 * Credit type information can help capture
   borrower financial habits and credit exposure patterns.

# 5 ) Overdue and Debt Analysis



## 5.1 Credit Overdue Distribution

```python

# CREDIT OVERDUE DISTRIBUTION
plt.figure(figsize=(12,5))
sns.histplot(bureau['CREDIT_DAY_OVERDUE'],bins=50,kde=True)
plt.xlim(0, 500)
plt.title('Credit Day Overdue Distribution')
plt.xlabel('Days Overdue')
plt.ylabel('Frequency')
plt.show()
```



 * The distribution is highly right-skewed,
   with most borrowers having zero overdue days.

* Only a small portion of borrowers
   show high overdue periods,
  indicating potential repayment risk.

 * Extreme overdue values may act as strong
   indicators of financial distress
 and default probability.

* CREDIT_DAY_OVERDUE can be useful   for identifying risky borrower behavior.

## 5.2 Credit Amount Distribution

```python
# CREDIT AMOUNT DISTRIBUTION
plt.figure(figsize=(12,5))
sns.histplot(bureau['AMT_CREDIT_SUM'],bins=50,kde=True)
plt.xlim(0, 3000000)
plt.title('Distribution of Credit Amount')
plt.xlabel('Credit Amount')
plt.ylabel('Frequency')
plt.show()
```

# 7 ) Correlation Analysis



## 7.1 Numerical Feature Correlation

```python
# NUMERICAL FEATURE CORRELATION
# Selected numerical features
corr_features = ['AMT_CREDIT_SUM','AMT_CREDIT_SUM_DEBT','AMT_CREDIT_SUM_OVERDUE','AMT_CREDIT_MAX_OVERDUE','CNT_CREDIT_PROLONG','DAYS_CREDIT','DAYS_CREDIT_ENDDATE','DAYS_CREDIT_UPDATE']

# Correlation matrix
corr_matrix = bureau[corr_features].corr()

# Plot heatmap
plt.figure(figsize=(12,8))
sns.heatmap(corr_matrix,annot=True,fmt='.2f',cmap='RdYlGn',center=0)
plt.title('Bureau Numerical Feature Correlation')
plt.show()
```



 * AMT_CREDIT_SUM and AMT_CREDIT_SUM_DEBT
   show strong positive correlation,
   indicating borrowers with larger credits
   generally carry higher debt.

 * DAYS_CREDIT and DAYS_CREDIT_UPDATE
   also exhibit strong correlation,
   reflecting consistent credit timeline behavior.

 * Overdue-related variables have weak correlations
   with most financial features,
 suggesting independent repayment patterns.
