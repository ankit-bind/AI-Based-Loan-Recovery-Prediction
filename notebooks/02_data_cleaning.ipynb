# 2 ) **application_train Dataset Cleaning**

```python
# LIBRARY IMPORT

# Data manipulation

import pandas as pd
import numpy as np

# Ignore warnings

import warnings
warnings.filterwarnings('ignore')

# Display settings

pd.set_option('display.max_columns',None)

pd.set_option('display.max_rows',100)
```

## 2.1 **Load Dataset**

```python
# LOAD APPLICATION TRAIN DATASET
application_train = pd.read_csv('application_train.csv')

# Dataset shape

print(f"Dataset Shape : {application_train.shape}")

# Preview dataset

application_train.head()
```

## 2.2 **Missing Value Handling**

```python
# MISSING VALUE ANALYSIS

# Calculate missing percentage

missing_percent = (application_train.isnull().mean()* 100)

# Create dataframe

missing_df = pd.DataFrame({ 'Missing Percentage':missing_percent})

# Sort values

missing_df = missing_df.sort_values(by='Missing Percentage',ascending=False)

# Display top missing features

missing_df.head(20)
```

```python
# Features with more than 30% missing values

high_missing = missing_df[missing_df['Missing Percentage'] > 30]

high_missing
```

## 2.3  **Missing Value Treatment**

```python
# REMOVE HIGH MISSING VALUE FEATURES

# Features with more than 60% missing values

high_missing_cols = missing_df[missing_df['Missing Percentage'] > 60].index

# Display columns

print('High Missing Features:\n')

print(list(high_missing_cols))
```

```python
# AVG / MODE / MEDI missing values comparison

cols = [

    'COMMONAREA_AVG',
    'COMMONAREA_MODE',
    'COMMONAREA_MEDI',

    'LIVINGAPARTMENTS_AVG',
    'LIVINGAPARTMENTS_MODE',
    'LIVINGAPARTMENTS_MEDI',

    'YEARS_BUILD_AVG',
    'YEARS_BUILD_MODE',
    'YEARS_BUILD_MEDI'

]

application_train[cols].isnull().mean() * 100

application_train[cols].corr()
```

```python
application_train[cols].corr()
```

```python
floor_cols = [

    'FLOORSMIN_AVG',
    'FLOORSMIN_MODE',
    'FLOORSMIN_MEDI'
]

application_train[floor_cols].corr()
```

## 2.4   **REMOVE REDUNDANT FEATURES**

```python
# REMOVE REDUNDANT FEATURES

remove_cols = [

    'COMMONAREA_MODE',
    'COMMONAREA_MEDI',

    'NONLIVINGAPARTMENTS_MODE',
    'NONLIVINGAPARTMENTS_MEDI',

    'LIVINGAPARTMENTS_MODE',
    'LIVINGAPARTMENTS_MEDI',

    'FLOORSMIN_MODE',
    'FLOORSMIN_MEDI',

    'YEARS_BUILD_MODE',
    'YEARS_BUILD_MEDI',

    'FONDKAPREMONT_MODE'

]

# Drop columns

application_train.drop(columns=remove_cols,inplace=True)

# Updated shape

print(f"Updated Dataset Shape : {application_train.shape}")
```

```python
# REMOVE ADDITIONAL HIGH-MISSING REDUNDANT FEATURES


remove_cols = [

    'LANDAREA_MODE',
    'LANDAREA_MEDI',

    'BASEMENTAREA_MODE',
    'BASEMENTAREA_MEDI',

    'NONLIVINGAREA_MODE',
    'NONLIVINGAREA_MEDI',

    'ELEVATORS_MODE',
    'ELEVATORS_MEDI',

    'APARTMENTS_MODE',
    'APARTMENTS_MEDI',

    'ENTRANCES_MODE',
    'ENTRANCES_MEDI',

    'LIVINGAREA_MODE',
    'LIVINGAREA_MEDI',

    'FLOORSMAX_MODE',
    'FLOORSMAX_MEDI',

    'YEARS_BEGINEXPLUATATION_MODE',
    'YEARS_BEGINEXPLUATATION_MEDI',

    'HOUSETYPE_MODE',
    'EMERGENCYSTATE_MODE'

]

# Drop columns safely

application_train.drop(columns=remove_cols,inplace=True,errors='ignore')

# Updated dataset shape

print(f"Updated Dataset Shape : {application_train.shape}")
```

## 2.5   **OWN_CAR_AGE Missing Value Handling**

```python
# HANDLE OWN_CAR_AGE MISSING VALUES

# Missing values before handling

print( "Missing Values Before :",application_train['OWN_CAR_AGE'].isnull().sum())

# Fill missing values with 0

application_train['OWN_CAR_AGE'] = application_train['OWN_CAR_AGE'].fillna(0)

# Missing values after handling

print("Missing Values After :",application_train['OWN_CAR_AGE'].isnull().sum())
```

## 2.6   **DAYS_EMPLOYED Anomaly Handling**

```python
# HANDLE DAYS_EMPLOYED ANOMALIES

# Create anomaly flag

application_train['DAYS_EMPLOYED_ANOMALY'] = (application_train['DAYS_EMPLOYED'] == 365243).astype(int)

# Replace anomaly values with NaN

application_train['DAYS_EMPLOYED'] = application_train['DAYS_EMPLOYED'].replace(365243,np.nan)

# Check anomaly count

print("Anomaly Count :",application_train['DAYS_EMPLOYED_ANOMALY'].sum())
```

```python
application_train.shape
```

```python
# DAYS_EMPLOYED_ANOMALY Count
application_train['DAYS_EMPLOYED_ANOMALY'].value_counts()
```

## 2.7  **Duplicate Record Removal**

```python
# REMOVE DUPLICATE RECORDS

# Dataset shape before removal

print(f"Shape Before : {application_train.shape}")

# Remove duplicates

application_train.drop_duplicates( inplace=True)


```

```python
# Dataset shape after removal

print(f"Shape After : {application_train.shape}")
```

## 2.8  **XNA Handling**

```python
# CODE_GENDER XNA fix — 4 rows only, 2 minute kaam
application_train['CODE_GENDER'] = application_train[
    'CODE_GENDER'
].replace('XNA', 'F')  # F is mode
```

## 2.9   **Data Type Optimization**

```python
# MEMORY USAGE BEFORE OPTIMIZATION
memory_before = (application_train.memory_usage().sum()/ 1024**2)
print(f"Memory Usage Before : {memory_before:.2f} MB")
```

```python
# OPTIMIZE DATA TYPES

# Float columns

float_cols = application_train.select_dtypes(include='float64').columns

# Integer columns

int_cols = application_train.select_dtypes(include='int64').columns

# Convert float64 → float32

application_train[float_cols] = application_train[float_cols].astype('float32')

# Convert int64 → int32

application_train[int_cols] = application_train[int_cols].astype('int32')
```

```python
# MEMORY USAGE AFTER OPTIMIZATION

memory_after = (application_train.memory_usage().sum()/ 1024**2)

print(f"Memory Usage After : {memory_after:.2f} MB")

# Memory reduction percentage

reduction = ((memory_before - memory_after)/ memory_before) * 100

print(f"Memory Reduced By : {reduction:.2f}%")
```

## 2.10   **Outlier Treatment**

```python
# OUTLIER TREATMENT — 99th PERCENTILE CAPPING

cols_to_cap = [
    'AMT_INCOME_TOTAL',
    'AMT_CREDIT',
    'AMT_ANNUITY',
    'AMT_GOODS_PRICE'
]

for col in cols_to_cap:

    # 99th percentile upper cap
    upper = application_train[col].quantile(0.99)

    # Before
    before_max = application_train[col].max()

    # Cap — only upper side, no lower cap for amounts
    application_train[col] = application_train[col].clip(upper=upper)

    # After
    after_max = application_train[col].max()

    print(f'{col:<25} | Before: {before_max:>15,.0f} | After: {after_max:>12,.0f}')

print('\nOutlier capping completed ✓')
```

## 2.11 **Save cleaned application_train dataset**

```python
# Save cleaned application_train dataset

application_train.to_csv( 'application_train_cleaned.csv',index=False)

print('application_train_cleaned.csv saved successfully!')
```

# 3 ) **bureau Dataset Cleaning**

## 3.1  **Load Dataset**

```python
# LOAD BUREAU DATASET

bureau = pd.read_csv('bureau.csv')

# Dataset shape

print(f"Dataset Shape : {bureau.shape}")

# Preview dataset

bureau.head()
```

## 3.2   **High Missing Feature Inspection**

```python
# BUREAU MISSING VALUE ANALYSIS

# Calculate missing percentage

bureau_missing_percent = (bureau.isnull().mean()* 100)

# Create dataframe

bureau_missing_df = pd.DataFrame({'Missing Percentage':bureau_missing_percent})

# Sort values

bureau_missing_df = bureau_missing_df.sort_values(by='Missing Percentage',ascending=False)

# Display missing values

bureau_missing_df
```

```python
# HIGH MISSING VALUE FEATURES

# Features with more than 30% missing values

bureau_high_missing = bureau_missing_df[bureau_missing_df['Missing Percentage'] > 30]

# Display high missing features

bureau_high_missing
```

## 3.3   **Duplicate Record Check**

```python
# DUPLICATE RECORD CHECK

# Total duplicate rows

duplicate_count = bureau.duplicated().sum()

print(f"Duplicate Records : {duplicate_count}")
```

## 3.4  **Data Type Optimization**

```python
# MEMORY USAGE BEFORE OPTIMIZATION

memory_before = (bureau.memory_usage().sum()/ 1024**2)

print(f"Memory Usage Before : {memory_before:.2f} MB")
```

```python
# OPTIMIZE NUMERICAL DATA TYPES


# Float columns

float_cols = bureau.select_dtypes(include='float64').columns

# Integer columns

int_cols = bureau.select_dtypes(include='int64').columns

# Convert float64 → float32

bureau[float_cols] = bureau[float_cols].astype('float32')

# Convert int64 → int32

bureau[int_cols] = bureau[int_cols].astype('int32')
```

```python
# MEMORY USAGE AFTER OPTIMIZATION


memory_after = ( bureau.memory_usage().sum()/ 1024**2)

print(f"Memory Usage After : {memory_after:.2f} MB")

# Memory reduction percentage

reduction = ((memory_before - memory_after)/ memory_before) * 100

print(f"Memory Reduced By : {reduction:.2f}%")
```

## 3.5  **Basic Bureau Aggregation**

```python
# ============================================================
# BASIC BUREAU AGGREGATION
# ============================================================

bureau_agg = bureau.groupby('SK_ID_CURR').agg({

    'AMT_CREDIT_SUM': ['sum', 'mean'],

    'AMT_CREDIT_SUM_DEBT': ['sum', 'mean'],

    'CREDIT_DAY_OVERDUE': ['max'],

    'SK_ID_BUREAU': ['count']

})

# Flatten column names

bureau_agg.columns = ['_'.join(col)for col in bureau_agg.columns]

# Reset index

bureau_agg.reset_index( inplace=True)

bureau_agg.head()
```

## 3.6 **Save cleaned bureau dataset**

```python
# Save cleaned bureau dataset

bureau.to_csv('bureau_cleaned.csv',index=False)

print('bureau_cleaned.csv saved successfully!')
```

## 3.7   **MERGE BUREAU AGGREGATED FEATURES**

```python

# MERGE BUREAU AGGREGATED FEATURES

application_train = application_train.merge(bureau_agg,on='SK_ID_CURR',how='left')

# Fill missing bureau values with 0

bureau_cols = [col for col in bureau_agg.columns if col != 'SK_ID_CURR']

application_train[bureau_cols] = application_train[bureau_cols].fillna(0)

print('Bureau merge completed successfully!')

print(f'Final Dataset Shape : {application_train.shape}')
```
